## Data Type Conversion

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])
```

```python
df.info()
```

## Feature Engineering

```python
df["Order Date"]
```

```python
df["Year"] = df["Order Date"].dt.year
df["Month"] = df["Order Date"].dt.month
df["Quarter"] = df["Order Date"].dt.quarter
```

```python
df["Shipping Days"] = (
    df["Ship Date"] - df["Order Date"]
).dt.days
```

```python
df.head()
```

## Feature Engineering 1

New features were created from the Order Date column to support time-based analysis during exploratory data analysis.

Features Created:
- Year
- Month
- Quarter
- Shipping Days

```python
(df["Ship Date"] < df["Order Date"]).sum()
```

##Validating Numeric Values

```python
print("Negative Sales:", (df["Sales"] < 0).sum())
print("Invalid Quantity:", (df["Quantity"] <= 0).sum())
print("Invalid Discount:", ((df["Discount"] < 0) | (df["Discount"] > 1)).sum())
```

##Feature Engineering 2
These features are created to make profitability and discounting easier to interpret and analyze.
They provide additional business-focused insights beyond the existing Sales, Profit, and Discount variables.
They can help identify patterns between sales, profitability, and discounting across different products or categories.


Features Created:
* Profit Margin
* Discount Percentage


```python
df["Profit Margin"] = (
    df["Profit"] / df["Sales"]
) * 100
```

```python
df["Discount Rate"] = df["Discount"] * 100
df.head()
```
